# Spotify Dataset — Initial EDA

این نوت‌بوک برای بررسی اولیه‌ی CSV واقعی است؛ قبل از preprocessing، خروجی‌ها را تحلیل کنید.

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path('../data/raw/spotify_tracks.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Shape:', df.shape)
df.info()
display(df.isna().sum().sort_values(ascending=False))
print('Duplicate rows:', df.duplicated().sum())
print('Unique track IDs:', df['track_id'].nunique())
print('Unique genres:', df['track_genre'].nunique())
```

```python
display(df['track_genre'].value_counts().rename_axis('genre').to_frame('count'))
display(df.describe(include='all').T)
```
